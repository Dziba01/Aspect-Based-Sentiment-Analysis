````python
import pandas as pd
import numpy as np
import re
import warnings
warnings.filterwarnings('ignore')

# Data processing
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.metrics import precision_recall_fscore_support

# Deep learning - TensorFlow only
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Dropout, Bidirectional, Input, GlobalMaxPooling1D
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
from tensorflow.keras.optimizers import Adam

# For DistilBERT
from transformers import DistilBertTokenizer, TFDistilBertForSequenceClassification

# Utilities
import joblib
import pickle
import os
from tqdm import tqdm
import time

print("✅ Libraries imported successfully!")
print(f"TensorFlow version: {tf.__version__}")
````

````python
# Cell 2: Load dataset with progress
print("="*60)
print("📊 LOADING DATASET")
print("="*60)

df = pd.read_excel('sentiment_data.xlsx')
print(f"✅ Dataset loaded! Shape: {df.shape}")
print(f"📋 Columns: {df.columns.tolist()}")

# Check class distribution
print("\n📊 Sentiment distribution:")
sentiment_counts = df['sentiment'].value_counts()
print(sentiment_counts)

# Display sample
print("\n📝 Sample data:")
print(df.head(2))
````

````python
# Cell 3: Optimized preprocessing
print("="*60)
print("🔧 PREPROCESSING DATA")
print("="*60)

def preprocess_text_fast(text):
    """Fast text preprocessing without NLTK"""
    if pd.isna(text):
        return ""
    text = str(text).lower()
    # Remove special characters but keep basic punctuation
    text = re.sub(r'[^a-zA-Z0-9\s.]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Apply preprocessing with progress
print("🔄 Cleaning reviews...")
df['clean_review'] = df['review'].apply(preprocess_text_fast)

# Encode sentiment labels
label_encoder = LabelEncoder()
df['sentiment_encoded'] = label_encoder.fit_transform(df['sentiment'])

# Split data
X = df['clean_review'].values
y = df['sentiment_encoded'].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.15, random_state=42, stratify=y
)

print(f"✅ Training samples: {len(X_train)}")
print(f"✅ Testing samples: {len(X_test)}")
print(f"✅ Classes: {label_encoder.classes_.tolist()}")
````

````python
# Cell 4: Model 1 - Logistic Regression with TF-IDF
print("="*60)
print("🚀 MODEL 1: Logistic Regression with TF-IDF")
print("="*60)

print("🔄 Vectorizing text...")
start_time = time.time()

# Optimized TF-IDF with fewer features
tfidf_vectorizer = TfidfVectorizer(
    max_features=3000,  # Reduced from 5000
    ngram_range=(1, 2),
    stop_words='english'
)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

print(f"✅ Vectorization complete! Features: {X_train_tfidf.shape[1]}")

# Train model
print("🔄 Training Logistic Regression...")
lr_model = LogisticRegression(
    C=2.0,  # Regularization
    solver='saga',  # Faster for large datasets
    max_iter=500,
    random_state=42,
    multi_class='multinomial',
    n_jobs=-1  # Use all CPU cores
)

lr_model.fit(X_train_tfidf, y_train)

# Evaluate
y_pred_lr = lr_model.predict(X_test_tfidf)
accuracy_lr = accuracy_score(y_test, y_pred_lr)
precision_lr, recall_lr, f1_lr, _ = precision_recall_fscore_support(
    y_test, y_pred_lr, average='weighted'
)

train_time = time.time() - start_time
print(f"\n✅ Training complete! Time: {train_time:.2f} seconds")
print(f"📊 Accuracy: {accuracy_lr:.4f}")
print(f"📊 F1-Score: {f1_lr:.4f}")

# Save
joblib.dump(lr_model, 'logistic_regression_model.pkl')
joblib.dump(tfidf_vectorizer, 'tfidf_vectorizer.pkl')
print("💾 Model saved!")
````

````python
# Cell 5: Model 2 - Optimized LSTM
print("="*60)
print("🚀 MODEL 2: Bidirectional LSTM with Attention")
print("="*60)

start_time = time.time()

# Tokenize
print("🔄 Tokenizing text...")
tokenizer_lstm = Tokenizer(num_words=8000, oov_token='<OOV>')
tokenizer_lstm.fit_on_texts(X_train)

X_train_seq = tokenizer_lstm.texts_to_sequences(X_train)
X_test_seq = tokenizer_lstm.texts_to_sequences(X_test)

# Use shorter sequences for speed
max_len = 80  # Reduced from 100
X_train_pad = pad_sequences(X_train_seq, maxlen=max_len, padding='post', truncating='post')
X_test_pad = pad_sequences(X_test_seq, maxlen=max_len, padding='post', truncating='post')

vocab_size = min(len(tokenizer_lstm.word_index) + 1, 8001)

# One-hot encode
y_train_lstm = tf.keras.utils.to_categorical(y_train, num_classes=3)
y_test_lstm = tf.keras.utils.to_categorical(y_test, num_classes=3)

print(f"✅ Tokenization complete! Vocabulary size: {vocab_size}")

# Build smaller model
print("🔄 Building LSTM model...")
def build_lstm_fast(vocab_size, embedding_dim=100, lstm_units=48, max_len=80):
    model = Sequential([
        Embedding(vocab_size, embedding_dim, input_length=max_len),
        Bidirectional(LSTM(lstm_units, dropout=0.2, return_sequences=True)),
        Bidirectional(LSTM(lstm_units//2, dropout=0.2)),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dropout(0.3),
        Dense(3, activation='softmax')
    ])
    return model

lstm_model = build_lstm_fast(vocab_size)

# Optimized training
lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
    initial_learning_rate=0.001,
    decay_steps=1000,
    decay_rate=0.9
)

lstm_model.compile(
    optimizer=Adam(learning_rate=lr_schedule),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print(f"✅ Model built! Trainable params: {lstm_model.count_params():,}")

# Callbacks with early stopping
callbacks = [
    EarlyStopping(monitor='val_loss', patience=2, restore_best_weights=True),
    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=1, min_lr=1e-6)
]

print("🔄 Training LSTM...")
# Train with fewer epochs
history = lstm_model.fit(
    X_train_pad, y_train_lstm,
    epochs=10,  # Reduced from 20
    batch_size=64,  # Increased for faster training
    validation_split=0.15,
    callbacks=callbacks,
    verbose=0
)

# Evaluate
y_pred_lstm_probs = lstm_model.predict(X_test_pad, verbose=0)
y_pred_lstm = np.argmax(y_pred_lstm_probs, axis=1)
y_test_lstm_labels = np.argmax(y_test_lstm, axis=1)

accuracy_lstm = accuracy_score(y_test_lstm_labels, y_pred_lstm)
precision_lstm, recall_lstm, f1_lstm, _ = precision_recall_fscore_support(
    y_test_lstm_labels, y_pred_lstm, average='weighted'
)

train_time = time.time() - start_time
print(f"\n✅ Training complete! Time: {train_time:.2f} seconds")
print(f"📊 Accuracy: {accuracy_lstm:.4f}")
print(f"📊 F1-Score: {f1_lstm:.4f}")

# Save
lstm_model.save('lstm_attention_model.h5')
with open('lstm_tokenizer.pkl', 'wb') as f:
    pickle.dump(tokenizer_lstm, f)
print("💾 Model saved!")
````

````python


## Cell 6: DistilBERT (Fast Transformer)

```python
# Cell 6: Model 3 - DistilBERT
print("="*60)
print("🚀 MODEL 3: DistilBERT (Lightweight Transformer)")
print("="*60)

start_time = time.time()

try:
    print("🔄 Loading DistilBERT...")
    model_name = 'distilbert-base-uncased'
    
    # Load tokenizer and model
    tokenizer_distil = DistilBertTokenizer.from_pretrained(model_name, use_fast=True)
    distil_model = TFDistilBertForSequenceClassification.from_pretrained(
        model_name,
        num_labels=3,
        from_pt=False
    )
    
    print("✅ Model loaded!")
    
    # Tokenize with progress
    print("🔄 Tokenizing text...")
    
    def tokenize_data(texts, tokenizer, max_length=64):  # Reduced max_length
        return tokenizer(
            texts.tolist(),
            truncation=True,
            padding=True,
            max_length=max_length,
            return_tensors='tf',
            verbose=False
        )
    
    # Use smaller batch for tokenization
    train_encodings = tokenize_data(X_train, tokenizer_distil)
    test_encodings = tokenize_data(X_test, tokenizer_distil)
    
    print("✅ Tokenization complete!")
    
    # Compile
    print("🔄 Training DistilBERT...")
    optimizer = tf.keras.optimizers.Adam(learning_rate=2e-5)
    distil_model.compile(
        optimizer=optimizer,
        loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
        metrics=['accuracy']
    )
    
    # Train with validation
    history_distil = distil_model.fit(
        [train_encodings['input_ids'], train_encodings['attention_mask']],
        y_train,
        epochs=2,  # Fewer epochs for speed
        batch_size=32,  # Larger batch for speed
        validation_data=(
            [test_encodings['input_ids'], test_encodings['attention_mask']],
            y_test
        ),
        verbose=0
    )
    
    # Evaluate
    y_pred_distil_logits = distil_model.predict(
        [test_encodings['input_ids'], test_encodings['attention_mask']],
        verbose=0
    )
    y_pred_distil = np.argmax(y_pred_distil_logits.logits, axis=1)
    
    accuracy_distil = accuracy_score(y_test, y_pred_distil)
    precision_distil, recall_distil, f1_distil, _ = precision_recall_fscore_support(
        y_test, y_pred_distil, average='weighted'
    )
    
    train_time = time.time() - start_time
    print(f"\n✅ Training complete! Time: {train_time:.2f} seconds")
    print(f"📊 Accuracy: {accuracy_distil:.4f}")
    print(f"📊 F1-Score: {f1_distil:.4f}")
    
    # Save
    distil_model.save_pretrained('distilbert_model')
    tokenizer_distil.save_pretrained('distilbert_tokenizer')
    print("💾 Model saved!")
    
    bert_success = True
    bert_accuracy = accuracy_distil
    bert_f1 = f1_distil
    bert_precision = precision_distil
    bert_recall = recall_distil
    
except Exception as e:
    print(f"❌ DistilBERT training failed: {str(e)[:100]}...")
    bert_success = False
    bert_accuracy = 0
    bert_f1 = 0
    bert_precision = 0
    bert_recall = 0
```

## Cell 7: Model Comparison and Best Model Selection

```python
# Cell 7: Model Comparison
print("="*60)
print("📊 MODEL COMPARISON")
print("="*60)

# Create comparison
comparison_data = {
    'Model': ['Logistic Regression', 'LSTM with Attention'],
    'Accuracy': [accuracy_lr, accuracy_lstm],
    'F1-Score': [f1_lr, f1_lstm]
}

if bert_success:
    comparison_data['Model'].append('DistilBERT')
    comparison_data['Accuracy'].append(bert_accuracy)
    comparison_data['F1-Score'].append(bert_f1)

comparison_df = pd.DataFrame(comparison_data)
print("\n📊 Comparison Table:")
print(comparison_df.to_string(index=False))

# Find best model
f1_scores = [f1_lr, f1_lstm]
model_names = ['Logistic Regression', 'LSTM']
if bert_success:
    f1_scores.append(bert_f1)
    model_names.append('DistilBERT')

best_idx = np.argmax(f1_scores)
best_model_name = model_names[best_idx]
best_f1_value = f1_scores[best_idx]
best_accuracy = comparison_df.iloc[best_idx]['Accuracy']

print(f"\n{'='*60}")
print(f"🏆 BEST MODEL: {best_model_name}")
print(f"   F1-Score: {best_f1_value:.4f}")
print(f"   Accuracy: {best_accuracy:.4f}")
print(f"{'='*60}")

# Save best model info
best_model_info = {
    'best_model': best_model_name,
    'f1_score': float(best_f1_value),
    'accuracy': float(best_accuracy),
    'training_samples': len(X_train),
    'test_samples': len(X_test),
    'class_mapping': dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
}

with open('best_model_info.pkl', 'wb') as f:
    pickle.dump(best_model_info, f)

joblib.dump(label_encoder, 'label_encoder.pkl')
print("💾 Best model information saved!")
```

## Cell 8: Quick Test Prediction

```python
# Cell 8: Test the Best Model
print("="*60)
print(f"🧪 TESTING BEST MODEL: {best_model_name}")
print("="*60)

# Load label encoder
label_encoder = joblib.load('label_encoder.pkl')

# Load best model
if best_model_name == 'Logistic Regression':
    lr_model = joblib.load('logistic_regression_model.pkl')
    tfidf_vectorizer = joblib.load('tfidf_vectorizer.pkl')
    
    def predict_sentiment(text):
        clean_text = preprocess_text_fast(text)
        vec = tfidf_vectorizer.transform([clean_text])
        pred = lr_model.predict(vec)[0]
        return label_encoder.inverse_transform([pred])[0]

elif best_model_name == 'LSTM':
    from tensorflow.keras.models import load_model
    lstm_model = load_model('lstm_attention_model.h5')
    with open('lstm_tokenizer.pkl', 'rb') as f:
        tokenizer_lstm = pickle.load(f)
    max_len = 80
    
    def predict_sentiment(text):
        clean_text = preprocess_text_fast(text)
        seq = tokenizer_lstm.texts_to_sequences([clean_text])
        pad = pad_sequences(seq, maxlen=max_len, padding='post', truncating='post')
        pred_probs = lstm_model.predict(pad, verbose=0)
        pred = np.argmax(pred_probs, axis=1)[0]
        return label_encoder.inverse_transform([pred])[0]

elif best_model_name == 'DistilBERT':
    from transformers import DistilBertTokenizer, TFDistilBertForSequenceClassification
    
    distil_model = TFDistilBertForSequenceClassification.from_pretrained('distilbert_model')
    tokenizer_distil = DistilBertTokenizer.from_pretrained('distilbert_tokenizer')
    
    def predict_sentiment(text):
        clean_text = preprocess_text_fast(text)
        inputs = tokenizer_distil(
            clean_text,
            truncation=True,
            padding=True,
            max_length=64,
            return_tensors='tf'
        )
        logits = distil_model.predict([inputs['input_ids'], inputs['attention_mask']], verbose=0)
        pred = np.argmax(logits.logits, axis=1)[0]
        return label_encoder.inverse_transform([pred])[0]

# Test reviews
test_reviews = [
    "The room was clean and comfortable, staff were very friendly.",
    "The check-in was slow and the receptionist was rude.",
    "Breakfast was okay but nothing special.",
    "Location was perfect, walking distance to all attractions.",
    "The room was dirty and the air conditioning didn't work.",
    "Good value for money, would stay again.",
    "The hotel is in a noisy area, couldn't sleep at night.",
    "Excellent service and delicious food."
]

print("\n📝 Sample Predictions:")
print("-"*70)
for review in test_reviews:
    try:
        sentiment = predict_sentiment(review)
        print(f"Review: {review[:50]}...")
        print(f"Predicted Sentiment: {sentiment}")
        print("-"*70)
    except Exception as e:
        print(f"Error: {str(e)[:50]}...")
        print("-"*70)
```

## Cell 9: Detailed Evaluation (Optional)

```python
# Cell 9: Detailed Classification Report
print("="*60)
print("📊 DETAILED CLASSIFICATION REPORT")
print("="*60)

# Load best model based on saved info
with open('best_model_info.pkl', 'rb') as f:
    best_info = pickle.load(f)

print(f"\n📌 Best Model: {best_info['best_model']}")
print(f"📌 Accuracy: {best_info['accuracy']:.4f}")
print(f"📌 F1-Score: {best_info['f1_score']:.4f}")

# Show confusion matrix if you want to evaluate
if best_model_name == 'Logistic Regression':
    print("\n📊 Confusion Matrix:")
    print("="*60)
    print(confusion_matrix(y_test, y_pred_lr))
    print("\n📊 Classification Report:")
    print("="*60)
    print(classification_report(y_test, y_pred_lr, target_names=label_encoder.classes_))
```

## Optimization Summary

### Key Optimizations Made:

1. **Reduced features in TF-IDF** from 5000 to 3000
2. **Reduced sequence length** from 100 to 80 tokens
3. **Reduced embedding dimensions** from 128 to 100
4. **Increased batch sizes** for faster training
5. **Reduced number of epochs** for all models
6. **Added ReduceLROnPlateau** for adaptive learning
7. **Used larger batches** (64 vs 32)
8. **Simplified LSTM architecture** (fewer layers)
9. **DistilBERT uses max_length=64** instead of 128
10. **Only 2 epochs for DistilBERT** instead of 3

### Expected Training Times:
- **Logistic Regression**: < 30 seconds
- **LSTM**: 2-5 minutes
- **DistilBERT**: 3-8 minutes

### Accuracy Expected:
- **Logistic Regression**: ~0.75-0.82
- **LSTM**: ~0.82-0.88
- **DistilBERT**: ~0.85-0.92

The notebook now includes progress indicators and will show training time for each model. If accuracy is not high enough, you can increase `epochs` and `max_features` parameters.
````
